## Histogram Processing

```python
f = plt.imread(imgpath+'/pollen-lowcontrast.tif')
print(f.shape,f.dtype)
```

- histogram equalization

```python
plt.imshow(f, cmap = 'gray', vmin=0, vmax=255)
plt.axis('off')
plt.colorbar()
plt.show()
```

```python
plt.hist(f.flatten())
plt.show()
#범위를 더 넓혀 equalization 해야함
```

```python
def histogram_equalization(image):
    histogram, bins = np.histogram(image.flatten(), bins=256, range=[0,256])
    cdf = histogram.cumsum()
    cdf_min = cdf.min()
    cdf_max = cdf.max()
    cdf_normalized = (cdf - cdf_min) * 255 / (cdf_max - cdf_min)
    cdf_normalized = cdf_normalized.astype(np.uint8)
    equalized_image = cdf_normalized[image]
    return equalized_image
    #이거 교재의 코드와 매칭하면 좋다
```

```python
f_eq = histogram_equalization(f)
plt.hist(f_eq.flatten())
plt.show()
```

```python
plt.imshow(f_eq, cmap = 'gray', vmin=0, vmax=255)
plt.axis('off')
plt.colorbar()
plt.show()
```

### Exercise 1.
- 아래의 예는 local histogram equalization을 수행한다.
- 적절한 neighborhood size N을 찾아서 숨겨진 기호들을 모두 확인해보자.
- plt.imshow의 vim, vmax를 수정하여서도 숨겨진 기호들을 확인할 수 있으나, 본 실습에서는 vmax=255로 고정하여 histogram processing을 통해 숨겨진 기호들을 확인하자.

```python
N = 10
f_eq_local = np.zeros_like(f)
for x in np.arange(0,256,N):
  for y in np.arange(0,256,N):
    f_eq_local[x:x+N,y:y+N] = histogram_equalization(f[x:x+N,y:y+N])

plt.imshow(f_eq_local,vmin=0,vmax=255,cmap='gray')
plt.axis('off')
plt.show()
```

## 2D convolution

```python
import scipy.signal as signal
from skimage import io
```

```python
f = io.imread('https://press.hufs.ac.kr/data/file/gift/6f5a7159a604fe7de30498fd76c8741d_JIWi6Gj9_d6e43320ba8ee4ac452deadd5246a2d6db29e853.jpg',as_gray=True)[::2,::2] #2칸 건너서 다운샘플링 하겠다는 뜻 그래서 약 2분의 1 다운샘플링 함
print(f.shape,f.dtype)
```

```python
plt.imshow(f,cmap='gray')
plt.axis('off')
plt.show()
```

```python
#w = np.random.randn(3,3)
#w는 커널의 크기
w = np.array([[-1,0,1],
              [-2,0,2],
              [-1,0,1]])
#이건 세로 방향의 변화가 강조되는 거, 가로방향의 edge를 강조하려면??
print(w)
```

```python
sz0,sz1 = f.shape
y = np.zeros_like(f)
for i in np.arange(1,sz0-1):
  for j in np.arange(1,sz1-1):
    y[i,j] = np.sum(f[i-1:i+2,j-1:j+2]*w[::-1,::-1])
#f랑 w 곱해 다 더하는 것 사실 가중평균
#위의 랜덤수가 돌릴때마다 달라지는데 그게 달라질 때마다 이 셀이 어떻게 달라지는지 확인해보기
#커널이 어떻게 적용되냐에따라 다른 필터의 결과가 나옴

plt.figure(dpi=150)
plt.subplot(1,2,1)
plt.imshow(f,cmap='gray')
plt.axis('off')
plt.title('original')
plt.subplot(1,2,2)
plt.imshow(y,cmap='gray')
plt.axis('off')
plt.title('filtered')
plt.show()
```

```python
y = signal.convolve2d(f,w,mode='same')
#모드가 same이라 padding이 있어야
y.shape
plt.figure(dpi=150)
plt.subplot(121)
plt.imshow(f,cmap='gray')
plt.axis('off')
plt.title('original')
plt.subplot(122)
plt.imshow(y,cmap='gray')
plt.axis('off')
plt.title('filtered')
plt.show()
```

## lowpass filtering - Gaussian kernel

```python
N = 9
std = 4
s = np.arange(-N,N+1,1)
t = np.arange(-N,N+1,1)
S,T = np.meshgrid(s,t)
wg = np.exp(-(S**2+T**2)/(2*std*std))
wg = wg/np.sum(wg)

plt.figure(figsize=(7,7))
plt.imshow(wg,cmap='gray')
plt.axis('off')
plt.colorbar()
plt.show()
```

```python
y_g = signal.convolve2d(f,wg,mode='same')

plt.figure(figsize=(16,9))
plt.subplot(1,3,1)
plt.imshow(f,cmap='gray')
plt.axis('off')
plt.title('original')
plt.subplot(1,3,2)
plt.imshow(y_g,cmap='gray')
plt.axis('off')
plt.title('filtered')
plt.subplot(1,3,3)
plt.imshow(y_g-f,cmap='gray',vmin=-.01,vmax=.01)
plt.axis('off')
plt.title('difference')
plt.show()
```

```python
s = np.arange(-13,14,1)
t = np.arange(-13,14,1)
S,T = np.meshgrid(s,t)

#시그마 1,3,5,7 값 바꿔가면서
for std in np.array([1,3,5,7]):
  wg = np.exp(-(S**2+T**2)/(2*std*std))
  wg = wg/np.sum(wg)

  y_g = signal.convolve2d(f,wg,mode='same')

  plt.figure(figsize=(16,9))
  plt.subplot(1,3,1)
  plt.imshow(f,cmap='gray')
  plt.axis('off')
  plt.title('original')
  plt.subplot(1,3,2)
  plt.imshow(y_g,cmap='gray')
  plt.axis('off')
  plt.title(f'filtered - sigma:{std}')
  plt.subplot(1,3,3)
  plt.imshow(y_g-f,cmap='gray',vmin=-.01, vmax=.01)
  plt.axis('off')
  plt.title('difference')
  plt.show()
```

## Sobel (1st order diff.)

```python
sobel1 = np.array([[-1,-2,-1],[0,0,0],[1,2,1]])
sobel2 = np.array([[-1,0,1],[-2,0,2],[-1,0,1]])
print(sobel1)
print(sobel2)
```

```python
g1 = signal.convolve2d(f,sobel1,mode='same')
g2 = signal.convolve2d(f,sobel2,mode='same')
G = np.sqrt(g1**2+g2**2)

plt.figure(figsize=(18,6))
plt.subplot(141)
plt.imshow(f,cmap='gray')
plt.colorbar()
plt.axis('off')
plt.title('original')
plt.subplot(142)
plt.imshow(g1,cmap='RdBu_r',vmin=-128,vmax=128)
plt.colorbar()
plt.axis('off')
plt.title('Gy')
plt.subplot(143)
plt.imshow(g2,cmap='RdBu_r',vmin=-128,vmax=128)
plt.colorbar()
plt.axis('off')
plt.title('Gx')
plt.subplot(144)
plt.imshow(G,cmap='gray',vmin=0,vmax=255)
plt.colorbar()
plt.axis('off')
plt.title('magnitude of gradient')
plt.show()
```
